#### 4.1.1. Handling Missing Price Values

The first step is to handle the missing values in the price-related columns.

At first we can deal with missing prices.

###  Missing discount %
Next, we check the entries where both original_price and discounted_price are available, but discount_percent is missing.

For these entries, the missing discount percentage can be calculated using:

$$ \text{discount\_percent} = \frac{\text{original\_price} - \text{discounted\_price}}{\text{original\_price}} \times 100 $$

Or equivalently:

$$ \text{discount\_percent} = 100 - \left( \frac{\text{discounted\_price}}{\text{original\_price}} \times 100 \right) $$

```python
nykaa_df[
    nykaa_df["original_price"].notna() & 
    nykaa_df["discounted_price"].notna() &  
    nykaa_df["discount_percent"].isna()]["product_name"].count()
```

#### Observation

There are **51 entries** where `original_price` and `discounted_price` are available, but `discount_percent` is missing.


#### Decision

For these 51 entries, the missing `discount_percent` is calculated using the available `original_price` and `discounted_price` values.


```python
null_disc_percent = nykaa_df["original_price"].notna() & nykaa_df["discounted_price"].notna() &  nykaa_df["discount_percent"].isna()
#null_disc_percent
```

```python
nykaa_df.loc[null_disc_percent, "discount_percent"] = 100 - (
    nykaa_df.loc[null_disc_percent, "discounted_price"] / 
    nykaa_df.loc[null_disc_percent, "original_price"]
) * 100
```

### 4.1.2. Handling Missing `original_price`

Next, we check the entries where `original_price` is missing. We compare these entries with `discounted_price` and `discount_percent` to understand why the values are missing and decide how they should be handled.


```python
nykaa_df[
    nykaa_df["original_price"].isna()
][["discounted_price", "discount_percent"]].count()
```

#### Observation

There are **7,141 entries** with a missing `original_price`. All of these entries have a `discounted_price`, while none of them have a `discount_percent`.


The dataset is checked again to confirm whether any entry has a missing `original_price` but still contains a `discount_percent`.


```python
nykaa_df[
    nykaa_df["original_price"].isna() &
    nykaa_df["discount_percent"].notna()
]
```

#### Observation

There are no entries with a missing `original_price` and an available `discount_percent`. Therefore, all **7,141 entries** are missing both `original_price` and `discount_percent`.


#### Decision

Since `discounted_price` is available for these entries, we set:

* `original_price` equal to `discounted_price`
* `discount_percent` equal to `0`

This means the product is treated as having no discount when the original price is not available.


```python
null_original_price = nykaa_df[
    "original_price"].isna() & nykaa_df[
        "discounted_price"].notna() &  nykaa_df[
        "discount_percent"].isna()
```

```python
nykaa_df.loc[null_original_price , "original_price"] = nykaa_df.loc[null_original_price , "discounted_price"]
nykaa_df.loc[null_original_price , "discount_percent"] = 0
```

### 5.4. Discount Percentage Cross-Validation

This step performs a crucial cross-validation check to ensure internal data consistency. We are verifying that the **scraped discount percentage** provided by the platform matches our own **calculated discount percentage** derived mathematically from the `original_price` and `discounted_price`. 

Because rounding variations are common on e-commerce platforms (e.g., a calculated 14.7% being displayed as 15%), we will allow a small, acceptable tolerance margin rather than demanding a perfect match. Large discrepancies, however, would flag potential data corruption or a fundamental misunderstanding of how the platform computes markdown values.

```python
calculated_discount = (
    (df_products['original_price'] - df_products['discounted_price'])
    / df_products['original_price'] * 100
).round()

(df_products['discount_percent'] != calculated_discount).sum()
```

```python
calculated_discount = (
    ((df_products['original_price'] - df_products['discounted_price'])
     / df_products['original_price'] * 100)
).round()

discount_check = df_products.copy()
discount_check['calculated_discount'] = calculated_discount

mismatches = discount_check[
    discount_check['discount_percent'] != discount_check['calculated_discount']
]

mismatches[[
    'product_name',
    'original_price',
    'discounted_price',
    'discount_percent',
    'calculated_discount'
]].head(10)
```

```python
mismatches['difference'] = (
    mismatches['discount_percent'] -
    mismatches['calculated_discount']
)

mismatches['difference'].value_counts().sort_index()
```

```python
discount_check['difference'] = (
    discount_check['discount_percent'] -
    discount_check['calculated_discount']
)

major_mismatch = discount_check[
    discount_check['difference'].abs() > 1
]

len(major_mismatch)
```

#### Observation: Uncovering Discount Discrepancies

To ensure our metrics tell a consistent story, I cross-validated the scraped discount percentages against a fresh mathematical calculation derived straight from the `original_price` and `discounted_price` columns. 

Here is what the data revealed:
* **Minor Rounding Differences:** The vast majority of slight variances (**136 records**) fell within a tiny ±1 percentage point window. These are completely normal and can be safely attributed to how the platform rounds display percentages on its front-end.
* **True Inconsistencies:** However, **20 products** showed significant discrepancies where the displayed discount completely drifted from the actual price values.

**The Fix:** Because the discount percentage is inherently a derived metric, I chose to recalculate it across the board using the raw price columns. This ensures absolute mathematical consistency across the entire dataset before we begin modeling or visualization.

### Decision: Preserving Both Discount Metrics

Instead of overwriting the original data, I decided to keep the raw, scraped discount percentage as-is and create a brand new column named `calculated_discount_percentage`. 

**Why this approach?**
* **Preserving Front-End Reality:** The scraped value reflects exactly what the customer saw on the website, which is invaluable for any future behavioral or conversion rate analysis.
* **Ensuring Mathematical Accuracy:** The new calculated column provides an uncompromised, exact mathematical metric derived strictly from our price points, giving us the clean data we need for modeling and financial analysis.

### 5.5. Engineering the Calculated Discount Percentage

The original scraped discount percentage is preserved in `discount_percent_scraped`.

The `discount_percent` column is then recalculated from the original and discounted prices using the standard discount formula:

$$
Discount\ \% =
\frac{Original\ Price - Discounted\ Price}
{Original\ Price}
\times 100
$$

This provides a mathematically consistent discount percentage for downstream analysis while preserving the original scraped value for reference.

```python
df_products['discount_percent_scraped'] = df_products['discount_percent']

df_products['discount_percent'] = (
    ((df_products['original_price'] - df_products['discounted_price'])
     / df_products['original_price'] * 100)
).round().astype(float)
```
